```python
import re
from pyspark.sql.functions import col

# Read CSV from Unity Catalog Volume
df = spark.read.option("headerRows", 1).excel('/Volumes/youtube_lakehouse/bronze/raw_landing/video_bronze.xlsx')

# Sanitize column names - replace characters invalid in Delta with underscores
df = df.toDF(*[re.sub(r'[ ,;{}()\n\t=]', '_', c) for c in df.columns])

# Cast timestamp_ntz columns to timestamp for Delta compatibility
df = df.withColumn("PublishedAt", col("PublishedAt").cast("timestamp")) \
       .withColumn("ScrapedAt", col("ScrapedAt").cast("timestamp"))

# Write to Silver table
df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("youtube_lakehouse.silver.fact_videos_history")

print(f"Successfully loaded {df.count()} rows into youtube_lakehouse.silver.fact_videos_history")
```
